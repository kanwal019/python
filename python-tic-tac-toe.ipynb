```python
import ipywidgets as widgets
from IPython.display import display, clear_output
```

```python
board = [""] * 9
current_player = "X"
game_over = False
```

```python
buttons = [widgets.Button(description="", layout=widgets.Layout(width="70px", height="70px", font_size="30px")) for _ in range(9)]
status = widgets.Label(value="Player X's turn")
reset_button = widgets.Button(description="Reset Game", button_style="warning")
```

```python
def check_winner(b):
    wins = [
        (0,1,2), (3,4,5), (6,7,8),  # rows
        (0,3,6), (1,4,7), (2,5,8),  # cols
        (0,4,8), (2,4,6)            # diagonals
    ]
    for i,j,k in wins:
        if board[i] and board[i] == board[j] == board[k]:
            return board[i]
    if all(board):
        return "Draw"
    return None
```

```python
def on_click(i):
    def handler(b):
        global current_player, game_over

        if game_over or board[i] != "":
            return
        
        board[i] = current_player
        buttons[i].description = current_player
        
        result = check_winner(board)
        if result:
            game_over = True
            if result == "Draw":
                status.value = "It's a draw!"
            else:
                status.value = f"Player {result} wins! 🎉"
            return

        current_player = "O" if current_player == "X" else "X"
        status.value = f"Player {current_player}'s turn"

    return handler
```

```python
for i, btn in enumerate(buttons):
    btn.on_click(on_click(i))
```

```python
def reset_game(b):
    global board, current_player, game_over
    board = [""] * 9
    current_player = "X"
    game_over = False
    status.value = "Player X's turn"
    for btn in buttons:
        btn.description = ""
```

```python
reset_button.on_click(reset_game)
```

```python
display(
    widgets.VBox([
        status,
        widgets.GridBox(buttons, layout=widgets.Layout(grid_template_columns="repeat(3, 70px)")),
        reset_button
    ])
)
```
