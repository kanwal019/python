# Python Functions

Note: Python uses snake casing i.e. `def name_of_function():`

```python
def print_hello():
    '''This function prints "Hello, World!" to the console.'''
    print("Hello")
    print("How are you?")
    print("I hope you are doing well!")

print_hello()  # Calling the function to print the message    
```

```python
def print_item(item):
    '''This function takes an item as input and prints it to the console.'''
    print(item)
```

```python
def add_numbers(a, b):
    '''This function takes two numbers as input and returns their sum.'''
    return a + b

x = 5
y = 3
z = add_numbers(x, y)  # Calling the function to add two numbers
print_item("The sum of {} and {} is {}".format(x, y, z))  # Calling the function to print the result
```
