# Python Functions

Note: Python uses snake casing i.e. `def name_of_function():`

```python
def print_hello():
    '''This function prints "Hello, World!" to the console.'''
    print("Hello, World!")

print_hello()  # Calling the function to print the message    
```

```python
def add_numbers(a, b):
    '''This function takes two numbers as input and returns their sum.'''
    return a + b

print(add_numbers(5, 3))  # Calling the function to add two numbers
```
