# Python Functions

Functions group reusable code under a name. Define a function with `def`, indent its body, and call it by adding parentheses to its name.

By convention, Python function names use `snake_case`, such as `name_of_function`.

## Defining and calling a function

The function below takes no arguments and prints three greeting messages. Defining it does not run its body; the call to `print_hello()` runs it.

Run the cells from top to bottom: later examples use functions defined earlier. The triple-quoted string at the start of a function body is its docstring; `help(print_hello)` displays that documentation.

```python
def print_hello():
    '''This function prints "Hello, World!" to the console.'''
    print("Hello")
    print("How are you?")
    print("I hope you are doing well!")

print_hello()  # Calling the function to print the message    
```

## Parameters and arguments

A parameter is a name in a function definition; an argument is a value supplied when calling the function. Here, `item` is a parameter of `print_item`. Calling `print_item("Hello")` passes `"Hello"` as its argument. This cell defines the function for use in the next example.

`item` has no default value, so the caller must supply it. Both `print_item("Hello")` and `print_item(item="Hello")` print `Hello`; calling `print_item()` raises `TypeError` because the required argument is missing.

```python
def print_item(item):
    '''This function takes an item as input and prints it to the console.'''
    print(item)
```

## Returning a value

Use `return` to send a result back to the caller. `add_numbers(a, b)` returns the sum of its two arguments, which is stored in `z`. The example then calls `print_item` to display the result.

Printing displays text; returning makes a value available for further use. A function without an explicit `return`, such as `print_item`, returns `None`.

With `x = 5` and `y = 3`, `z` is `8`, and the printed message is `The sum of 5 and 3 is 8`. A `return` statement also ends the current function call; statements after it in the same execution path do not run.

```python
def add_numbers(a, b):
    '''This function takes two numbers as input and returns their sum.'''
    return a + b

x = 5
y = 3
z = add_numbers(x, y)  # Calling the function to add two numbers
print_item("The sum of {} and {} is {}".format(x, y, z))  # Calling the function to print the result
```

## Default parameter values

A default value lets the caller omit an argument. In this version of `print_hello`, passing `"Alice"` uses that name, while calling `print_hello()` uses `"Default User"`. The f-strings insert the name into each greeting.

Running this cell replaces the earlier definition of `print_hello` with this version.

The two calls print three greetings each, separated by a blank line. You can also pass the name by keyword, for example `print_hello(name="Sam")`. This is a redefinition rather than a separate overload: after this cell runs, calls use the version with the optional `name` parameter. Rerunning the first definition restores the original version.

```python
def print_hello(name="Default User"):
    '''This function takes a name as input and prints a personalized greeting.'''
    print(f"Hello, {name}!")
    print(f"How are you, {name}?")
    print(f"I hope you are doing well, {name}!")

print_hello("Alice")  # Calling the function with a name argument
print("")  # Adding a blank line for better readability
print_hello()  # Calling the function without an argument, using the default value    
```

```python
result = add_numbers(10, 20)  # Calling the function to add two numbers
type_of_result = type(result)  # Getting the type of the result
print_item(f"The result of adding 10 and 20 is {result}, and its type is {type_of_result}.")  # Calling the function to print the result and its type

result = print_item(result) # Calling the function to print the result of the addition
type_of_result = type(result)  # Getting the type of the result
print_item(f"The result of print_item is {result}, and its type is {type_of_result}.")  # Calling the function to print the result and its type
```
