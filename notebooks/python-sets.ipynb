# Working with Sets in Python

The previous cells demonstrate basic operations on a Python set named `myset`:

- **Cell 0**: Initializes an empty set `myset = set()`.
- **Cell 1**: Displays the empty set (output: `set()`).
- **Cell 2**: Adds the element `1` to the set using `myset.add(1)`.
- **Cell 3**: Displays the set after adding `1` (output: `{1}`).
- **Cell 4**: Adds elements `2` and `3` to the set.
- **Cell 5**: Displays the set (output: `{1, 2, 3}`).
- **Cell 6**: Attempts to add `2` again (which is already in the set) and displays it. Sets automatically handle duplicates, so the set remains `{1, 2, 3}`.

Sets in Python are unordered collections of unique elements, useful for membership testing and eliminating duplicates.

```python
myset.add(4)
myset.add(5)
myset
```
