# Python File Open Modes

When you open a file in Python using `open(path, mode)`, the `mode` string controls how the file is opened. Common modes include:

- `r` : read (default). File must exist.
- `w` : write. Creates a new file or truncates an existing file.
- `a` : append. Opens for writing, adding data to the end of the file if it exists.
- `x` : exclusive creation. Fails if the file already exists.

You can combine these with:

- `b` : binary mode (e.g., `rb`, `wb`, `ab`). Use for non-text data.
- `t` : text mode (default). Can be explicit (`rt`, `wt`, etc.).

You can also allow both reading and writing:

- `r+` : read/write, file must exist (does not truncate).
- `w+` : read/write, truncates the file or creates it.
- `a+` : read/append, creates file if it does not exist; writes go to end unless you seek.
- `x+` : create and open for read/write, fails if file exists.


```python
with open("../files/test.txt", "r") as myfile:
    content = myfile.read()
    print(content)
# The file is automatically closed when exiting the 'with' block
```

```python
with open("../files/test.txt", mode="w") as myfile:
        myfile.write("This is a new line added to the file.\n")
        myfile.write("This is another line added to the file.\n")
# The 'w' mode will overwrite the existing content of the file, so only the new lines will be present in the file after this block.
with open("../files/test.txt", "a+") as mynewfile:
    mynewfile.write("This line is added to the end of the file using append mode.\n")
    mynewfile.seek(0) 
# Move the file pointer back to the beginning of the file to read the content after appending.
    content = mynewfile.read()
    print(content)
```

```python
with open("../files/temp_new.txt", "a") as myfile:
    myfile.write("This is a new line added to the new file.\n")
# This will create a new file named temp_new.txt and add the line to it. If the file already exists, it will append the line to the existing content.
```

```python
with open("../files/temp.txt", "x") as myfile:
    myfile.write("This is a new line added to the new file.\n")
# The 'x' mode will create a new file and write to it. If the file already exists, it will raise a FileExistsError.
```
