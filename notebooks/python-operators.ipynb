## 1. The Six Comparison Operators

For the built-in values used here, comparisons return a Boolean: `True` or `False`.

| Operator | Meaning | Example | Result |
| --- | --- | --- | --- |
| `==` | Equal to | `5 == 5` | `True` |
| `!=` | Not equal to | `5 != 3` | `True` |
| `<` | Less than | `5 < 3` | `False` |
| `<=` | Less than or equal to | `5 <= 5` | `True` |
| `>` | Greater than | `5 > 3` | `True` |
| `>=` | Greater than or equal to | `5 >= 6` | `False` |

`=` assigns a value; `==` compares values. Python uses `!=`, not `<>`, for inequality.

```python
a = 10
b = 5

print("a == b:", a == b)
print("a != b:", a != b)
print("a < b: ", a < b)
print("a <= b:", a <= b)
print("a > b: ", a > b)
print("a >= b:", a >= b)
print("Result type:", type(a > b))
```

## 2. Equality and Boundary Values

`<` and `>` exclude equality. `<=` and `>=` include it. This distinction matters at thresholds such as a passing score or minimum age.

Compatible numeric types can be compared by value: `10 == 10.0` is `True`, even though their types differ. `True` and `False` behave numerically as `1` and `0`, so equality does not necessarily mean two values have the same type.

```python
passing_score = 50
for score in (49, 50, 51):
    print(f"Score {score}: > 50 -> {score > passing_score}, >= 50 -> {score >= passing_score}")

print("10 == 10.0:", 10 == 10.0)
print("-3 < -1:", -3 < -1)
print("True == 1:", True == 1)
print("False == 0:", False == 0)
```

## 3. Comparing Strings

Strings are compared lexicographically: Python compares corresponding characters by Unicode code point until it finds a difference. If one string is a prefix of the other, the shorter string comes first. Comparisons are case-sensitive and are not locale-aware dictionary sorting.

Numeric strings are still strings: `"10" < "2"` is `True`. Convert valid numeric text with `int()` or `float()` when you want a numeric comparison. For a case-insensitive comparison, `casefold()` provides a normalized form without changing the original strings.

```python
print('"apple" == "Apple":', "apple" == "Apple")
print('"apple" < "banana":', "apple" < "banana")
print('"app" < "apple":', "app" < "apple")
print('"Z" < "a":', "Z" < "a")
print('"10" < "2":', "10" < "2")
print('int("10") < int("2"):', int("10") < int("2"))
print("Case-insensitive equality:", "Python".casefold() == "PYTHON".casefold())
```

## 4. Comparing Different Types

Equality between a number and a string returns `False`; Python does not automatically convert the string to a number. Ordering an integer and a string raises `TypeError`.

Not every numeric type supports ordering either: complex numbers support equality but do not support `<`, `<=`, `>`, or `>=`. The example below catches an expected error so later cells can still run.

```python
print('5 == "5":', 5 == "5")
print('5 == int("5"):', 5 == int("5"))

try:
    print(5 < "10")
except TypeError as error:
    print("Expected TypeError:", error)
```

## 5. Chained Comparisons

Use `lower <= value <= upper` to test an inclusive range. Chaining is equivalent to combining adjacent comparisons with `and`, except that an intermediate expression is evaluated only once. Evaluation stops as soon as a comparison fails.

`0 < temperature < 100` excludes both endpoints. `0 <= temperature <= 100` includes them.

`a == b == c` means all three are equal. However, `a != b != c` only checks `a != b` and `b != c`; it does not check `a != c`.

```python
for temperature in (-1, 0, 50, 100, 101):
    print(f"{temperature:3}: strict -> {0 < temperature < 100}, inclusive -> {0 <= temperature <= 100}")

age = 25
print("Chained:", 18 <= age <= 65)
print("With and:", 18 <= age and age <= 65)

print("1 != 2 != 1:", 1 != 2 != 1)  # True, despite the matching endpoints.
print("All pairs different:", 1 != 2 and 2 != 1 and 1 != 1)
```

## 6. Combining Conditions

- `and`: both conditions must be true.
- `or`: at least one condition must be true.
- `not`: reverses the truth value.

Comparisons bind more tightly than these logical operators; among logical operators, `not` comes before `and`, then `or`. Parentheses make your intended grouping clear.

`and` and `or` short-circuit and return one of their operands, so their results are not always Booleans. Here, their operands are comparisons or Boolean values. Use `and` and `or` for these conditions, rather than the bitwise operators `&` and `|`.

```python
age = 20
has_ticket = True
can_enter = age >= 18 and has_ticket
print("Can enter:", can_enter)

temperature = 35
outside_comfort_range = temperature < 18 or temperature > 28
print("Outside comfort range:", outside_comfort_range)
print("Within comfort range:", not outside_comfort_range)

# The second condition is skipped when the denominator is zero.
denominator = 0
print("Safe division condition:", denominator != 0 and 10 / denominator > 2)
```

## 7. Equality Versus Identity

`==` compares values; `is` checks whether two references point to the same object. Two different lists can contain equal values. An alias refers to the same list.

Use `is None` or `is not None` to check for `None`. Use `==` for string and numeric value comparisons; object reuse can make identity tests on those values misleading.

```python
first = [1, 2, 3]
second = [1, 2, 3]
alias = first

print("Equal contents:", first == second)
print("Same object:", first is second)
print("Alias is same object:", first is alias)

result = None
print("No result yet:", result is None)
print("Result available:", result is not None)
```

## 8. Floating-Point Comparisons

Many decimal fractions cannot be represented exactly as binary floating-point numbers. As a result, `0.1 + 0.2 == 0.3` is `False`.

When approximate equality is appropriate, use `math.isclose()`. Choose tolerances for your problem; a positive absolute tolerance is particularly useful when comparing to zero. Exact equality is still appropriate when exact matching is what you intend.

```python
import math

total = 0.1 + 0.2
print("Computed total:", total)
print("Exact equality:", total == 0.3)
print("Approximately equal:", math.isclose(total, 0.3, rel_tol=1e-9))
print("Close to zero:", math.isclose(1e-12, 0.0, abs_tol=1e-9))
```
