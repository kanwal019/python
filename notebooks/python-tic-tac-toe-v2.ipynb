### Validating User Input

Converting a response with `int()` does not ensure that it is a valid board position. The first cell repeats the direct conversion so you can explore invalid input; an error stops that cell, but you can still run the later function definition.

`user_choice()` adds a retry loop with two checks:

- `choice.isdigit()` checks for digit-only text. Blank responses, signs, spaces, and decimal points fail this check.
- Membership in `range(0, 10)` checks whether the converted number is between `0` and `9`, inclusive.

**The prompt says `0-10`, but the function accepts only `0-9`** because the stop value of `range()` is excluded. This is a separate validation example; its accepted values are not restricted to the three column indices used above.

Define the function, then run `user_choice()`. Try `hello`, then `10`, then `2` to see both retry messages followed by an accepted integer. Some Unicode characters pass `isdigit()` but cannot be converted by `int()`, so this example does not handle every possible input.

````python
# Read and convert a column index: 0 is left, 1 is center, and 2 is right.
# Non-integer text raises ValueError; int() alone does not check the column bounds.
# An IndexError can occur later when an out-of-range index is used to access a row.
position_index = int(input('Choose an index position: '))
````

````python
def user_choice():
    '''Prompt repeatedly until an integer from 0 through 9 is accepted, then return it.'''
    # Start with invalid text so the loop prompts at least once.
    choice ='WRONG'
    # The stop value is excluded: this accepts 0 through 9, despite the 0-10 prompt.
    # This range is a validation example, not a column index for the three-square rows.
    acceptable_range = range(0, 10)
    within_range = False
    # Retry until the response passes both the digit check and the range check.
    while choice.isdigit() == False or not within_range:
        choice = input('Please enter a number(0-10): ')
        # isdigit() rejects empty text, signs, decimal points, and whitespace.
        # Some Unicode digits pass isdigit() but cannot be converted by int().
        if choice.isdigit() == False:
            print('Sorry, that is not a number. Please try again.')
        else:
            # Convert digit text and check membership in the allowed range.
            if int(choice) in acceptable_range:
                within_range = True
            else:
                within_range = False
                print('Sorry, that number is not in the acceptable range. Please try again.')
    # Return a number rather than the string received from input().
    return int(choice)
````

### Simple User Interaction

Combine the same ideas in a small list-editing exercise. This example uses `game_list = [0, 1, 2]`, independently of the three-row board above.

| Function | Purpose |
| --- | --- |
| `display_game(game_list)` | Print the current list. |
| `position_choice()` | Keep prompting until the user enters exactly `0`, `1`, or `2`, then return that index as an integer. |

Run each function definition before its example call. These example calls prompt for input immediately. Replacement values remain strings, and `replacement_choice()` changes the supplied list in place; it does not restrict entries to `X` or `O` or check whether a position is occupied.

The final cell in this section resets the list and starts the interaction loop:

1. Display the current list.
2. Choose a position and enter its replacement string.
3. Display the updated list.
4. Enter `Y` to repeat or `N` to stop.

For example, choose `1`, enter `X`, and then enter `N` to see `[0, 'X', 2]` and exit after one update. Run that loop cell again to start over with `[0, 1, 2]`.

````python
game_list = [0,1,2]
````

````python
def display_game(game_list):
    '''This function displays the current game list.'''
    print("Here is the current game list:")
    print(game_list)
````

````python
def position_choice():
    '''Prompt the user to pick a position from the game list.'''
    choice = 'wrong'
    while choice not in ['0','1','2']:
        choice = input("Pick a position to replace (0, 1, or 2): ")
        if choice not in ['0','1','2']:
            print("Sorry, invalid choice!")
    return int(choice)
````

````python
def replacement_choice(game_list, position):
    '''Prompt the user to enter a replacement value for the game list.'''
    user_placement = input("Type a string to place at position {}: ".format(position))
    game_list[position] = user_placement
    return game_list
````

````python
replacement_choice(game_list, position_choice())
````

````python
def gameon_choice():
    '''Prompt the user to decide whether to keep playing.'''
    choice = 'wrong'
    while choice not in ['Y','N']:
        choice = input("Do you want to keep playing? (Y or N): ")
        if choice not in ['Y','N']:
            print("Sorry, I didn't understand that. Please enter Y or N.")
    if choice == 'Y':
        return True
    else:
        return False
````

````python
gameon_choice()
````

````python
game_on = True
game_list = [0,1,2]

while game_on:
    display_game(game_list)
    position = position_choice()
    game_list = replacement_choice(game_list, position)
    display_game(game_list)
    game_on = gameon_choice()
````

## Build the two-player game

The ten steps below use a single board list with ten entries. Index `0` is unused; indices `1-9` represent playable squares, and `' '` represents an empty square. This board is separate from the row lists and list-editing exercise above.

Define the functions in order, then run step 10 to play. Test cells let you inspect individual functions before combining them. The display uses `IPython.display.clear_output()` to replace the previous board in Jupyter; widgets are not required.

### Step 1: Display the board

Define `display_board(board)` to print positions `1-9` in three rows. The current implementation numbers squares from left to right, starting at the top:

```text
1 | 2 | 3
4 | 5 | 6
7 | 8 | 9
```

Use a ten-item list so index `0` can remain unused. Each call clears the previous cell output before printing the board.

````python
from IPython.display import clear_output

def display_board(board):
    '''This function displays the current board.'''
    clear_output()
    print("Here is the current board:")
    print(board[1] + ' | ' + board[2] + ' | ' + board[3])
    print(board[4] + ' | ' + board[5] + ' | ' + board[6])
    print(board[7] + ' | ' + board[8] + ' | ' + board[9])
````

### Step 2: Assign player markers

`player_input()` asks Player 1 to choose `X` or `O`, converts the response to uppercase, and returns `(player1, player2)` with the other marker assigned to Player 2.

Enter `X` or `O` when trying this function. In the current implementation, `return` is inside the loop, so an invalid response is returned immediately rather than prompting again. The loop therefore does not yet enforce a valid marker.

````python
def player_input():
    '''This function allows the player to choose their marker.'''
    marker = ''
    while marker != 'X' and marker != 'O':
        marker = input("Player 1: Do you want to be X or O? ").upper()
        player1 = marker
        if player1 == 'X':
            player2 = 'O'
        else:
            player2 = 'X'
        return (player1, player2)
````

### Step 3: Place a marker

`place_marker(board, marker, position)` changes the supplied board in place by assigning `marker` at `position`. Use positions `1-9` for playable squares.

This function does not check bounds or whether a square is occupied. In the complete game, `player_choice()` selects an available square before this function is called.

````python
def place_marker(board, marker, position):
    '''This function places the player's marker on the board.'''
    board[position] = marker
````

### Step 4: Check for a win

`win_check(board, mark)` returns `True` when the given marker fills any of the three rows, three columns, or two diagonals; otherwise, it returns `False`.

Step 10 calls this after placing each move and before checking for a draw, so a winning final move takes precedence over a full board.

````python
def win_check(board, mark):
    '''This function checks if a player has won the game.'''
    return ((board[1] == board[2] == board[3] == mark) or # across the top
            (board[4] == board[5] == board[6] == mark) or # across the middle
            (board[7] == board[8] == board[9] == mark) or # across the bottom
            (board[1] == board[4] == board[7] == mark) or # down the left side
            (board[2] == board[5] == board[8] == mark) or # down the middle
            (board[3] == board[6] == board[9] == mark) or # down the right side
            (board[1] == board[5] == board[9] == mark) or # diagonal
            (board[3] == board[5] == board[7] == mark))   # diagonal
````

### Step 5: Choose the starting player

`choose_first()` uses `random.randint(0, 1)` to return either `'Player 1'` or `'Player 2'`. Starting order is independent of marker choice, so either `X` or `O` can go first. Step 10 chooses a starter again for each new round.

````python
import random

def choose_first():
    '''This function uses the random module to randomly decide which player goes first.'''
    flip = random.randint(0,1)
    if flip == 0:
        return 'Player 1'
    else:
        return 'Player 2'
````

### Step 6: Check an available square

`space_check(board, position)` returns `True` when the selected entry is a single space (`' '`), and `False` when it contains a marker. Call it with a playable position from `1` through `9`; it does not validate the position itself.

````python
def space_check(board, position):
    '''This function checks if a space on the board is available.'''
    return board[position] == ' '
````

### Step 7: Check for a full board

`full_board_check(board)` checks positions `1-9` using `space_check()`. It returns `False` as soon as it finds an empty square, or `True` if all playable squares are occupied. Index `0` is ignored.

A full board is a draw only if the latest move did not win.

````python
def full_board_check(board):
    '''This function checks if the board is full.'''
    for i in range(1,10):
        if space_check(board, i):
            return False
    return True
````

### Step 8: Choose an available position

`player_choice(board)` reads a position, converts it with `int()`, and retries while the number is outside `1-9` or the square is occupied. It returns an available position as an integer.

The range check runs before the board lookup, so out-of-range integers are rejected without indexing the board. Blank input, words, and decimal text such as `3.5` still raise `ValueError`; this function does not catch conversion errors. Enter whole numbers while playing.

````python
def player_choice(board):
    '''This function prompts the player to choose a position on the board.'''
    position = 0
    while position not in range(1,10) or not space_check(board, position):
        position = int(input('Choose your next position: (1-9) '))
    return position
````

### Step 9: Ask to play again

`replay()` converts the response to lowercase and returns `True` only when it equals `'yes'`. Enter `Yes` to start another round or `No` to exit. Letter case is ignored, but `Y` and responses containing extra spaces do not count as `yes`. Other responses return `False` without a retry.

````python
def replay():
    '''This function asks the player if they want to play again.'''
    choice = input("Do you want to play again? Enter Yes or No: ").lower()
    return choice == 'yes'
````

### Step 10: Run the game

Run the function definitions from steps 1-9 before this cell. The outer loop starts each round with an empty board, fresh marker assignments, and a randomly chosen first player. The `markers` dictionary associates each player with their marker, allowing both players to share one move loop.

For each turn:

1. Display the board, then announce the current player and marker, such as `Player 1's turn (X)`. The announcement comes after `display_board()` because that function clears the previous output.
2. Ask for an available position and place the current marker.
3. Check for a win, then a draw, displaying the final board and result when the round ends.
4. Switch players only if the round continues.

At the readiness prompt, enter `Yes` to begin or `No` to skip the round. This prompt strips surrounding spaces and accepts any response starting with `y`, ignoring case; blank input skips the round. After a win, a draw, or skipping, `replay()` asks whether to start again. Enter `Yes` to reset the board and repeat setup, or `No` to exit.

Use `X` or `O` for marker choice and whole-number positions `1-9`. The input limitations described in steps 2, 8, and 9 also apply here.

````python
print('Welcome to Tic Tac Toe v2!')

while True:
    # Start a fresh round and associate each player with their marker.
    the_board = [' '] * 10
    player1_marker, player2_marker = player_input()
    markers = {'Player 1': player1_marker, 'Player 2': player2_marker}
    turn = choose_first()
    print(f'{turn} will go first.')

    play_game = input('Are you ready to play? Enter Yes or No: ')
    game_on = play_game.strip().lower().startswith('y')

    while game_on:
        marker = markers[turn]
        display_board(the_board)
        # Display the turn after the board because display_board clears the output.
        print(f"{turn}'s turn ({marker})")
        position = player_choice(the_board)
        place_marker(the_board, marker, position)

        if win_check(the_board, marker):
            display_board(the_board)
            print(f'Congratulations! {turn} ({marker}) has won!')
            game_on = False
        elif full_board_check(the_board):
            display_board(the_board)
            print('The game is a draw!')
            game_on = False
        else:
            turn = 'Player 2' if turn == 'Player 1' else 'Player 1'

    if not replay():
        break
````
