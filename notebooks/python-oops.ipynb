## 2. Define a class and create an instance

The `Sample` class accepts a value through the parameter `my_parameter` and stores it in an instance attribute named `my_attribute`.

| Syntax | Meaning |
| --- | --- |
| `class Sample()` | Defines a class named `Sample`. |
| `__init__(self, my_parameter)` | Initializes an instance with the supplied value. |
| `self` | Refers to the instance being initialized or used. |
| `self.my_attribute = my_parameter` | Stores the parameter's value in the instance attribute `my_attribute`. |

A **parameter** is a name used to receive an argument inside a method. An **instance attribute** stores data on an object. Their names can differ: here, `my_parameter` receives the value, while `self.my_attribute` keeps it on the instance.

Calling `Sample("Sample text")` creates an instance and automatically calls `__init__` to initialize it. Python supplies `self`; the argument `"Sample text"` becomes the value of `my_parameter`.

The cells below use `type(my_sample)` to inspect the instance's class and `my_sample.my_attribute` to read its stored value, `"Sample text"`.

```python
class Sample():
    def __init__(self, my_parameter):
        self.my_attribute = my_parameter
```

```python
my_sample = Sample("Sample text")
```

```python
type(my_sample)
```

```python
my_sample.my_attribute
```

## 3. Define instance and class attributes

The `Dog` class defines a class attribute, `species = "mammal"`, directly in the class body. It also accepts three parameters and stores each value as an instance attribute:

| Parameter | Instance attribute | Value in the example |
| --- | --- | --- |
| `breed` | `self.breed` | `"Labrador"` |
| `name` | `self.name` | `"Sammy"` |
| `spots` | `self.spots` | `False` |

**Class versus instance attributes:** `species` belongs to the class and can be accessed as `Dog.species`. Instances can also read it as `my_dog.species` when they have no instance attribute with that name. In contrast, `breed`, `name`, and `spots` are assigned through `self` and stored separately on each instance.

Calling `Dog(breed="Labrador", name="Sammy", spots=False)` supplies **keyword arguments**, which identify each parameter by name. Python supplies `self` automatically.

The next cells read the attributes with dot notation:

- `my_dog.breed` returns `"Labrador"`.
- `my_dog.name` returns `"Sammy"`.
- `my_dog.spots` returns `False`.

Each instance stores its own values. To create another dog, supply all three arguments, for example `Dog(breed="Beagle", name="Milo", spots=True)`.

**Try it:** Create a second dog and compare its attributes with those of `my_dog`.

```python
class Dog():
    species = "mammal" # class attribute
    def __init__(self, breed, name, spots):
        self.breed = breed
        self.name = name
        self.spots = spots
    def bark(self, number=1):
        return "Woof! My name is {} and I am a {}. ".format(self.name, self.breed) * number   
```

```python
my_dog = Dog(breed="Labrador", name="Sammy", spots=False)
```

```python
my_dog.breed
```

```python
my_dog.name
```

```python
my_dog.spots
```

## 4. Understand attribute types and input validation

The next cell uses **positional arguments**: `Dog("Labrador", "Sammy", "No spots")`. These values map to `breed`, `name`, and `spots` in that order, and the assignment replaces the value of `my_dog` with a new instance.

The current `__init__` method stores each argument directly without checking its type. As a result, `my_dog.spots` returns the string `"No spots"`, even though the earlier example used the Boolean value `False`.

> **Input validation:** If `spots` must be a Boolean, add a check inside `__init__` before assigning `self.spots`. For example, `if not isinstance(spots, bool):` followed by `raise TypeError("spots must be a bool")` rejects values of other types. This validation is an exercise; the current class does not enforce it.

### Key takeaways

- A **class** defines a type; an **instance** stores its own attributes.
- A class attribute such as `Dog.species` provides a value that instances can read unless they override it with an instance attribute of the same name.
- `__init__` initializes attributes, and `self` refers to the current instance.
- Keyword arguments identify parameters by name; positional arguments follow parameter order.
- Dot notation accesses attributes such as `my_dog.name` and `my_dog.spots`.
- Assigning an attribute does not enforce its type. Add explicit validation when a parameter must accept only certain values.

**Try it:** Add Boolean validation to `Dog.__init__`, then compare what happens when you pass `False` and `"No spots"`.

```python
my_dog = Dog("Labrador", "Sammy", "No spots")
my_dog.spots 
```

## 6. Define and call an instance method

A **method** is a function defined in a class. The `Dog` class now includes `bark(self, number=1)`, which uses `self.name` and `self.breed` to build a message for the current dog.

The next cells create a new instance with `Dog("Husky", "Frankie", True)` and assign it to `my_dog`. Its `species` is still `"mammal"`, and its `name` is `"Frankie"`.

### Referencing a method versus calling it

| Expression | Result |
| --- | --- |
| `my_dog.bark` | Retrieves a bound method associated with `my_dog`; it does not run the method. |
| `my_dog.bark()` | Calls the method with the default `number=1` and returns one message. |
| `my_dog.bark(2)` | Returns the message repeated twice. |
| `my_dog.bark(number=3)` | Uses a keyword argument to repeat the message three times. |

Python supplies `self` automatically when you call `my_dog.bark()`. You only pass `number` if you want to override its default value.

### How the message is built

The expression `"Woof! My name is {} and I am a {}. ".format(self.name, self.breed)` inserts the dog's name and breed into the placeholders. Multiplying that string by `number` repeats it.

For Frankie, `my_dog.bark()` returns `"Woof! My name is Frankie and I am a Husky. "`, including a trailing space. The method **returns** the string; Jupyter displays it when the method call is the final expression in a cell.

**Try it:** Call `my_dog.bark(2)` and `my_dog.bark(number=3)`. Then create another dog and call its `bark()` method to see how the message uses that instance's attributes.

```python
my_dog = Dog("Husky", "Frankie", True)
```

```python
my_dog.species
```

```python
my_dog.name
```

### Inspect and invoke `bark`

The next two cells show the difference between accessing a method and calling it:

- `my_dog.bark` returns a **bound method** associated with the Frankie instance. Jupyter displays a representation of the method without executing its body.
- `my_dog.bark()` executes the method with the default `number=1` and returns `"Woof! My name is Frankie and I am a Husky. "`.

The parentheses trigger the call. Python supplies `self` automatically so the method can read this dog's `name` and `breed`.

```python
my_dog.bark
```

```python
my_dog.bark()
```
