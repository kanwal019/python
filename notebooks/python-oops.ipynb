# Object-Oriented Programming (OOP) in Python

Object-oriented programming groups data and behavior into **objects**. A **class** defines a type of object; an **instance** is an individual object of that class.

In this notebook, you will:

- Inspect the types of built-in objects with `type()`.
- Define a class and initialize its instances with `__init__`.
- Store and access instance attributes using dot notation.

## 1. Explore built-in objects

Python lists and sets are objects too. The next cells create a `list` and an empty `set`, then inspect their types with `type()`.

> **Tip:** Use `set()` to create an empty set. Empty braces (`{}`) create a dictionary.

```python
mylist = [1,2,3]
myset = set()
```

```python
type(mylist)
```

```python
type(myset)
```

## 2. Define a class and create an instance

The `Sample` class stores a piece of text in an instance attribute named `text`.

| Syntax | Meaning |
| --- | --- |
| `class Sample()` | Defines a class named `Sample`. |
| `__init__(self, text)` | Initializes an instance with the supplied `text` value. |
| `self` | Refers to the instance being initialized or used. |
| `self.text = text` | Stores the supplied value on that instance. |

Calling `Sample("Sample text")` creates an instance and automatically calls `__init__` to initialize it. Python supplies `self`; you pass only the `text` argument.

The cells below use `type(my_sample)` to inspect the instance's class and `my_sample.text` to read its stored value, `"Sample text"`.

```python
class Sample():
    def __init__(self, text):
        self.text = text
```

```python
my_sample = Sample("Sample text")
```

```python
type(my_sample)
```

```python
my_sample.text
```

## 3. Practice with a `Dog` class

The same pattern can represent a dog and its breed:

1. Define the `Dog` class with a `breed` parameter in `__init__`.
2. Store the value in the instance attribute `self.breed`.
3. Create `my_dog` with the keyword argument `breed="Labrador"`.
4. Read the value using `my_dog.breed`, which returns `"Labrador"`.

Each `Dog` instance can store its own breed. For example, `Dog(breed="Beagle")` would create another instance with a different value.

### Key takeaways

- A **class** defines a type; an **instance** is an object of that type.
- `__init__` initializes an instance's attributes.
- `self` gives a method access to the current instance.
- Dot notation, such as `my_dog.breed`, accesses an object's attributes.

**Try it:** Create a second dog with a different breed and inspect both dogs' `breed` attributes.

```python
class Dog():
    def __init__(self, breed):
        self.breed = breed
```

```python
my_dog = Dog(breed="Labrador")
```

```python
my_dog.breed
```
