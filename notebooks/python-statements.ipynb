# Conditional Statements in Python

Use conditional statements to decide which code runs based on a condition. This notebook introduces `if`, `elif`, and `else` through Boolean flags, location checks, and personalized greetings.

## Learning Goals

- Run a block only when a condition is true.
- Provide an alternative with `else`.
- Choose between several conditions with `elif`.
- Understand indentation, branch order, and case-sensitive comparisons.

**Getting started:** Run each example, then change its input and predict the next result. These examples require Python 3 and no additional packages.

## Conditional Syntax

```python
if condition:
    # Runs when the first condition is truthy.
    pass
elif another_condition:
    # Checked only if earlier conditions were falsy.
    pass
else:
    # Runs if none of the conditions matched.
    pass
```

Each branch header ends with a colon. Indentation defines the block; use four spaces per level consistently. Align `if`, `elif`, and `else` at the same level. `pass` is a placeholder statement that does nothing, useful when a block has not been implemented yet.

An `if` can stand alone. You can add zero or more `elif` branches and an optional final `else`. Unlike `if` and `elif`, `else` has no condition.

## 1. Boolean Conditions: `if` and `else`

An `if` statement tests the truth value of its condition. In this cell:

1. `if True` always runs its block, printing `This is a test.`.
2. With `hungry = True`, `if hungry` prints `I'm hungry.`.
3. After `hungry = False`, the final `if` block is skipped and its `else` prints `I'm not hungry.`.

These are three separate `if` statements, so each is evaluated independently. The `else` belongs only to the final `if`.

`if hungry:` is enough to test a Boolean flag; writing `if hungry == True:` is unnecessary. Conditions can also test other values: `None`, numeric zero, and empty strings or containers are falsy, while nonempty strings and containers are truthy. For example, the string `"False"` is truthy because it is not empty.

**Try it:** Change the second assignment to `hungry = True` and predict the last line. The earlier assignment is overwritten before the final condition is tested.

````python
if True:
    print("This is a test.")

hungry = True
if hungry:
    print("I'm hungry.")

hungry = False
if hungry:
    print("I'm hungry.")
else:
    print("I'm not hungry.")    
````

## 2. Choosing a Branch with `if`, `elif`, and `else`

`elif` means “else if.” Python checks the conditions from top to bottom and executes only the **first matching branch** in this chain. If none match, the `else` block runs.

The example uses `==` to compare string values. Remember that `=` assigns a value, while `==` checks equality.

| Value of `loc` | Printed message |
| --- | --- |
| `'Auto Shop'` | `Cars are cool!` |
| `'Bank'` | `Money is cool!` |
| `'Store'` | `Welcome to the store!` |
| `'Game'` or another unmatched value | `I don't know much about that location.` |

Here, `loc = 'Game'` does not match any listed location, so the fallback message is printed. String comparisons are case-sensitive: `'bank'` does not equal `'Bank'`.

**Try it:** Rerun the cell with each listed location and then with `'bank'`. Unlike several independent `if` statements, this chain selects at most one branch.

````python
loc = 'Game'

if loc == 'Auto Shop':
    print("Cars are cool!")
elif loc == 'Bank':
    print("Money is cool!")
elif loc == 'Store':
    print("Welcome to the store!")        
else:
    print("I don't know much about that location.")    
````

## 3. Personalized Greetings

This example applies the same branching pattern to a name. With `name = "Sam"`, the first comparison against `"Frankie"` is false, the next comparison is true, and the output is `Hello Sam!`. The remaining branches are skipped.

Try `"Frankie"`, `"George"`, and an unlisted name to exercise every branch. `"sam"` and `"Sam "` do not match `"Sam"`, because capitalization and whitespace affect equality.

When the intended behavior is to ignore surrounding whitespace and letter case, normalize the value before comparing it:

```python
normalized_name = name.strip().casefold()
if normalized_name == "sam":
    print("Hello Sam!")
```

`strip()` removes leading and trailing whitespace, and `casefold()` supports case-insensitive matching. These methods return new strings; the original `name` remains unchanged.

````python
name = "Sam"

if name == "Frankie":
    print("Hello Frankie!")
elif name == "Sam":
    print("Hello Sam!")
elif name == "George":
    print("Hello George!")
else:
    print("Hello, what's your name?")            
````

## Practice and Common Mistakes

Try these changes in the examples above:

4. Create a score example with `score >= 80`, then `score >= 50`, then `else`. Try 49, 50, 79, and 80. Put the higher threshold first so it is not hidden by the broader condition.

Check these details when a branch behaves unexpectedly:

- End every branch header with `:` and indent its body consistently.
- Use `==` for equality comparisons and `=` for assignments.
- Put more specific conditions before broader ones when they overlap.
- Place `else` last and do not give it a condition.
- For multiple alternatives, write `loc == "Bank" or loc == "Store"`, or use `loc in ("Bank", "Store")`. The expression `loc == "Bank" or "Store"` is always truthy because `"Store"` is a nonempty string.
