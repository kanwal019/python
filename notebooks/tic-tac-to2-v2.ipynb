### Validating User Input

```python
# Read and convert a column index: 0 is left, 1 is center, and 2 is right.
# Non-integer text raises ValueError; int() alone does not check the column bounds.
# An IndexError can occur later when an out-of-range index is used to access a row.
position_index = int(input('Choose an index position: '))
```

```python
def user_choice():
    '''Prompt repeatedly until an integer from 0 through 9 is accepted, then return it.'''
    # Start with invalid text so the loop prompts at least once.
    choice ='WRONG'
    # The stop value is excluded: this accepts 0 through 9, despite the 0-10 prompt.
    # This range is a validation example, not a column index for the three-square rows.
    acceptable_range = range(0, 10)
    within_range = False
    # Retry until the response passes both the digit check and the range check.
    while choice.isdigit() == False or not within_range:
        choice = input('Please enter a number(0-10): ')
        # isdigit() rejects empty text, signs, decimal points, and whitespace.
        # Some Unicode digits pass isdigit() but cannot be converted by int().
        if choice.isdigit() == False:
            print('Sorry, that is not a number. Please try again.')
        else:
            # Convert digit text and check membership in the allowed range.
            if int(choice) in acceptable_range:
                within_range = True
            else:
                within_range = False
                print('Sorry, that number is not in the acceptable range. Please try again.')
    # Return a number rather than the string received from input().
    return int(choice)
```

```python
# Run the function definition first; this call waits for input.
# Try invalid text, then 10, then 2 to see both retry messages and a valid result.
user_choice()
```

### Simple User Interaction

```python
game_list = [0,1,2]
```

```python
def display_game(game_list):
    '''This function displays the current game list.'''
    print("Here is the current game list:")
    print(game_list)
```

```python
display_game(game_list)
```

```python
def position_choice():
    '''Prompt the user to pick a position from the game list.'''
    choice = 'wrong'
    while choice not in ['0','1','2']:
        choice = input("Pick a position to replace (0, 1, or 2): ")
        if choice not in ['0','1','2']:
            print("Sorry, invalid choice!")
    return int(choice)
```

```python
position_choice()
```

```python
def replacement_choice(game_list, position):
    '''Prompt the user to enter a replacement value for the game list.'''
    user_placement = input("Type a string to place at position {}: ".format(position))
    game_list[position] = user_placement
    return game_list
```

```python
replacement_choice(game_list, position_choice())
```

```python
def gameon_choice():
    '''Prompt the user to decide whether to keep playing.'''
    choice = 'wrong'
    while choice not in ['Y','N']:
        choice = input("Do you want to keep playing? (Y or N): ")
        if choice not in ['Y','N']:
            print("Sorry, I didn't understand that. Please enter Y or N.")
    if choice == 'Y':
        return True
    else:
        return False
```

```python
gameon_choice()
```

```python
game_on = True
game_list = [0,1,2]

while game_on:
    display_game(game_list)
    position = position_choice()
    game_list = replacement_choice(game_list, position)
    display_game(game_list)
    game_on = gameon_choice()
```
