## Preparation

Run the cells in order to explore board display and update a square. Each list represents one row, and each item represents one square.

```python
# Display one row with numbers labeling its three squares.
print([1,2,3])
```

```python
# Display three rows to show the layout of all nine squares.
print([1,2,3])
print([4,5,6])
print([7,8,9])
```

```python
def display(row1, row2, row3):
    '''Print the board's top, middle, and bottom rows, one list per line.'''
    print(row1)
    print(row2)
    print(row3)
```

```python
# Create a separate list for each row; a space marks an empty square.
row1 = [' ', ' ', ' ']
row2 = [' ', ' ', ' ']
row3 = [' ', ' ', ' ']
# Display the empty board using the function defined above.
display(row1, row2, row3)
```

```python
# Run the previous cell first to create the board rows.
# Place X in the top-right square: row1 is the top row, and index 2 is its third square.
row1[2] = 'X'
# Display the board again to see the updated square.
display(row1, row2, row3)
```

## User Input

Run these cells in order after preparing the board. Each `input()` call pauses until you type a value and press Enter. Use integer text such as `4` for the conversion example, then choose `0`, `1`, or `2` to look up a square in the middle row.

```python
# Read text from the user; Jupyter displays the returned string without storing it.
input('Press enter a value...')
```

```python
# Store the response for the next cells; input() returns a string, even for digits.
# Enter a whole number such as 4 so the later int() conversion succeeds.
result = input('Press enter a value...')
```

```python
# Inspect the saved response and confirm that its type is str.
print(result)
print(type(result))
```

```python
# Convert the response to an integer, leaving the original result string unchanged.
# Blank input or text such as 'hello' or '3.5' raises ValueError.
result_int = int(result)
print(result_int)
print(type(result_int))
```

```python
# Read and convert a column index: 0 is left, 1 is center, and 2 is right.
# This cell converts the input but does not validate whether the index is in range.
position_index = int(input('Choose an index position: '))
print(position_index)
print(type(position_index))
```
