Run the cells in order to explore board display and update a square. Each list represents one row, and each item represents one square.

```python
# Display one row with numbers labeling its three squares.
print([1,2,3])
```

```python
# Display three rows to show the layout of all nine squares.
print([1,2,3])
print([4,5,6])
print([7,8,9])
```

```python
def display(row1, row2, row3):
    '''Print the board's top, middle, and bottom rows, one list per line.'''
    print(row1)
    print(row2)
    print(row3)
```

```python
# Create a separate list for each row; a space marks an empty square.
row1 = [' ', ' ', ' ']
row2 = [' ', ' ', ' ']
row3 = [' ', ' ', ' ']
# Display the empty board using the function defined above.
display(row1, row2, row3)
```

```python
# Run the previous cell first to create the board rows.
# Place X in the top-right square: row1 is the top row, and index 2 is its third square.
row1[2] = 'X'
# Display the board again to see the updated square.
display(row1, row2, row3)
```
