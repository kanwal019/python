# Tic Tac Toe v2

Preperation

```python
# Printing a row of a tic-tac-toe board
print([1,2,3])
```

```python
# Printing the 3 rows of a tic-tac-toe board
print([1,2,3])
print([4,5,6])
print([7,8,9])
```

```python
def display(row1, row2, row3):
    '''This function displays the rows of a tic-tac-toe board.'''
    print(row1)
    print(row2)
    print(row3)
```

```python
# Printing the 3 rows of a tic-tac-toe board using the display function
row1 = [' ', ' ', ' ']
row2 = [' ', ' ', ' ']
row3 = [' ', ' ', ' ']
display(row1, row2, row3)
```

```python
# Making a move on the tic-tac-toe board
row1[2] = 'X'
display(row1, row2, row3)
```
