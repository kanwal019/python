# Python Strings: Tutorial and Practice

Learn how to create and work with strings in Python through examples and hands-on exercises. Explore indexing, slicing, string methods, concatenation, and formatting, then practice extracting characters, transforming text, and building messages.

```python
# This code snippet will print the character at index 8 of the string 'Hello World'.
'Hello World'[8]
```

```python
# This code snippet will print the substring of 'tinker' starting from index 1 up to (but not including) index 4.
'tinker'[1:4]
```

```python
name= "Sam"
name[0] = 'P'
# The above code will raise an error because strings in Python are immutable, meaning you cannot change their characters directly. If you want to change the first character of the string, you can create a new string instead:
```

```python
name = "Pam"
last_letters = name[1:]
# The above code creates a new string 'Pam' by concatenating 'P' with the substring of 'name' starting from index 1 to the end. The variable 'last_letters' holds the substring 'am', and when we concatenate it with 'S', we get the final string 'Sam'.
'S' + last_letters
# The above code will output 'Sam' by concatenating the string 'S' with the substring 'am' stored in the variable 'last_letters'.
```

```python
letter = 'z'
letter * 10
# The above code will output 'zzzzzzzzzz' by repeating the string
```

```python
x = "Hello World!"
x.upper() # The above code will output 'HELLO WORLD!' by converting all characters in the string to uppercase.
```

```python
x.lower() # The above code will output 'hello world!' by converting all characters in the string to lowercase.
```

```python
x.split() # The above code will output ['Hello', 'World!'] by splitting the string into a list of words based on whitespace.
```

```python
x.split("o") # The above code will output ['Hell', ' W', 'rld!'] by splitting the string into a list of substrings based on the character 'o'.
```
