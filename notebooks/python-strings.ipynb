# Print Formatting with Strings

```python
print("This is a string {}".format("Hello, World!"))
# The above code will output 'This is a string Hello, World!' by using the format method to insert the string 'Hello, World!' into the placeholder {} in the original string.
```

```python
print("The {2} {1} {0}".format("fox", "brown", "quick"))
# The above code will output 'The quick brown fox' by using the format method to insert
```

```python
print("The {2} {2} {2}".format("fox", "brown", "quick"))
# The above code will output 'The quick quick quick' by using the format method to insert the string 'quick' into all three placeholders {} in the original string, as the index 2 corresponds to the third argument passed to the format method.
```

```python
print("The {q} {b} {f}".format(f="fox", b="brown", q="quick"))
# The above code will output 'The quick brown fox' by using the format method to insert the strings 'quick', 'brown', and 'fox' into the placeholders {q}, {b}, and {f} respectively in the original string.
```

### Float formatting follows "{value:width.precision f}"

```python
result = 100 / 777
result
```

```python
print("The result of 100 divided by 777 is: {}".format(result))
print("The result of 100 divided by 777 is: {:.2f}".format(result))
# The above code will output 'The result of 100 divided by 777 is: 0.13' by formatting the result to two decimal places using the format method.
print("The result of 100 divided by 777 is: {r:1.5f}".format(r=result))
# The above code will output 'The result of 100 divided by 777 is: 0.12869' by formatting the result to five decimal places using the format method with a specified width of 1.
```
