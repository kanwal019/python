## Creating Tuples and Converting Sequences

Both `()` and `tuple()` create an empty tuple. `tuple(iterable)` collects the items from an iterable into a tuple; here, converting `[1, 2, 3, 4, 5]` produces `(1, 2, 3, 4, 5)`. Conversely, `list(tpl)` creates a list containing the tuple's items.

**The comma makes a tuple:** `(42,)` is a one-item tuple, while `(42)` is simply the integer `42`. For multiple items, parentheses are often optional: `1, 2, 3` also creates a tuple.

Conversions do not modify the original container or deeply copy its items. Nested mutable objects remain shared. Use `len(tpl)` for the number of items and `value in tpl` to check membership.

````python
# Tuples in python

emp_tpl=()
print(emp_tpl)
print(type(emp_tpl))

tpl=tuple()
print(tpl)
print(type(tpl))

tpl=tuple([1,2,3,4,5])
print(tpl)
print(type(tpl))

lst=list(tpl)
print(lst)
print(type(lst))
````

## Indexing and Slicing

Tuples use the same indexing rules as lists. Index `0` selects the first item, and `-1` selects the last. An invalid individual index raises `IndexError`.

The slice syntax is `sequence[start:stop:step]`: start is included, stop is excluded, and the default step is `1`. A tuple slice returns a tuple. Out-of-range slice bounds are clipped rather than raising an error.

| Expression | Result | Explanation |
| --- | --- | --- |
| `elements_tpl[2:5]` | `('Jon', 'Jack', 1)` | Items at indices 2 through 4 |
| `elements_tpl[-1]` | `True` | The final item itself |
| `elements_tpl[-1:]` | `(True,)` | A one-item tuple |
| `elements_tpl[-1:3]` | `()` | A forward step cannot reach the stop |
| `elements_tpl[-1:-3]` | `()` | Negative indices alone do not reverse direction |
| `elements_tpl[3:-3]` | `('Jack', 1, 2, 3)` | Stops before the third item from the end |
| `elements_tpl[-3:-1]` | `(4, 5)` | Excludes the last item |

Omitting bounds includes the corresponding end of the sequence. `[::3]` selects every third item from the beginning; `[::-1]` reverses the order; `[::-3]` selects every third item from the end. A step of `0` raises `ValueError`. These operations leave the original tuple unchanged.

````python
# Accessing tuple elements, they are accessed in the same manner as lists

elements=["John", "Jane", "Jon", "Jack", 1, 2, 3, 4, 5, True]
print(elements)
print(type(elements))

elements_tpl=tuple(elements)
print(elements_tpl)
print(type(elements_tpl))

print(elements_tpl[2:])
print(elements_tpl[2:5])
print(elements_tpl[-1:])
print(elements_tpl[-1:3])
print(elements_tpl[-1:-3])
print(elements_tpl[3:-3])
print(elements_tpl[-3:-1])

print(elements_tpl[2:])
print(elements_tpl[2:5])
print(elements_tpl[5:])
print(elements_tpl[:5])
print(elements_tpl[::3])
print(elements_tpl[::-1])
print(elements_tpl[::-3])
````

## Concatenation and Repetition

`range(1, 20)` supplies integers 1 through 19, excluding 20. Despite its name, `num_lst` is a `range` object in this cell; `tuple(num_lst)` converts it to a 19-item tuple.

- `num_tpl + elements_tpl` concatenates two tuples, producing a 29-item tuple.
- `num_tpl * 4` repeats the entire sequence four times, producing 76 items. It does not multiply each numeric value by four.

Both operations produce tuple results without modifying the operands. Repetition reuses references to the items; it does not create independent copies of any nested mutable objects.

````python
# Insight into tuples 

num_lst=range(1,20)
num_tpl=tuple(num_lst)

mixed_tpl=num_tpl+elements_tpl
print(mixed_tpl)

tpl_multi=num_tpl*4
print(tpl_multi) 
````

## What Immutability Means

A tuple's item references cannot be replaced, added, or removed after creation. The commented assignment `num_tpl[1] = 99` would raise `TypeError` if uncommented. Tuples have no `append()`, `insert()`, or `remove()` methods.

The list and tuple below are separate containers. Replacing index 1 in `num_lst` produces `[1, 99, 3, 4, 5]`, while `num_tpl` remains `(1, 2, 3, 4, 5)`. Assigning a new tuple to the same variable is allowed: that rebinds the name rather than changing the original tuple.

**Immutability is not recursive.** A tuple can contain a mutable list:

```python
record = ('scores', [10, 20])
record[1].append(30)  # Allowed: changes the inner list.
# record[1] = [30]   # Not allowed: replaces a tuple item.
```

A tuple can be a dictionary key or set element only if all its items are hashable. A tuple containing a list is not hashable.

````python
# Tuples are immutable

num_lst=list([1,2,3,4,5])
num_tpl=tuple(num_lst)

num_lst[1]=99
print(num_lst)

# num_tpl[1]=99        # This fails as 'tuple' object does not support item assignment
print(num_tpl)
````

## Counting and Finding Items

Tuples provide two methods:

| Method | Behavior |
| --- | --- |
| `count(value)` | Returns the number of equal items, or `0` if there are none. |
| `index(value)` | Returns the first matching index; raises `ValueError` if there is no match. |

The previous cell changed the original `2` in `num_lst` to `99`. This cell then adds three occurrences of `2` and converts the updated list to `(1, 99, 2, 3, 4, 5, 2, 2)`. The printed count is therefore `3`, and the first matching index is `2`.

Both methods compare using equality. For example, `True` and `1` compare equal, so `(True, 1).count(1)` is `2`. Rerunning this cell adds more items to the list; restart and run from the beginning to reproduce the counts above.

````python
# Tuple methods

num_lst.append(2)
num_lst.append(2)
num_lst.insert(2,2)

num_tpl=tuple(num_lst)
print(num_tpl.count(2))
print(num_tpl.index(2))
````

## Packing and Unpacking

**Packing** groups comma-separated values into a tuple: `tpl = "John", 1, 2, 3, True`. **Unpacking** assigns those values to separate variables in order.

Without a starred target, the number of targets must match the number of items. Here, `i, j, k, l, m = tpl` assigns all five values; too many or too few targets raise `ValueError`.

A starred target collects the remaining items into a **list**, even when the source is a tuple:

```python
fst, *mid, lst = tpl
# fst: 'John'
# mid: [1, 2, 3]
# lst: True
```

At most one target at each unpacking level may be starred. It can receive an empty list, provided enough items exist for the other targets. In this cell, the name `lst` is reused for the final Boolean value; it no longer refers to the list created earlier.

````python
# Packing and unpacking; a starred target collects remaining items

tpl="John", 1, 2, 3, True
print(tpl)

i,j,k,l,m=tpl
print(i)
print(j)
print(k)
print(l)
print(m)

fst,*mid,lst=tpl
print(fst)
print(mid)
print(lst)
````

## Nested Tuples and Iteration

A tuple can contain other tuples. Apply indexing or slicing one step at a time:

- `tpl[0]` returns the first inner tuple: `(1, 2, 3, 4)`.
- `tpl[0][0:2]` selects that inner tuple, then its first two items: `(1, 2)`.
- `tpl[1:][1]` first keeps the last two inner tuples, then selects index 1 of that result: `('Alpha', 'Mike', 'Foxtrot')`. Here, `tpl[2]` accesses the same inner tuple directly.

The outer loop visits each inner tuple, and the inner loop visits its items. `print(item, end=" ")` keeps each group on one line; the final `print()` starts the next line.

Like lists, tuples support direct iteration and `enumerate()` when both an index and an item are needed. Use tuples for a fixed sequence of values and lists when the sequence needs to grow, shrink, or have its items replaced.

````python
# Nested tuples

tpl=((1,2,3,4),(6,7,8,9),("Alpha","Mike","Foxtrot"))
print(tpl[0])
print(tpl[0][0:2])
print(tpl[1:][1])

for sub in tpl:
    for item in sub:
        print(item,end=" ")
    print()    
````
